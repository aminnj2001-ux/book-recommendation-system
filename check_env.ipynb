# Development Environment Check

This notebook records the Python version, Python executable path, and installed versions of NumPy and Matplotlib. Run all cells before submitting.

```python
import sys
import platform
import importlib.metadata

print('Python version:', sys.version)
print('Python executable:', sys.executable)
print('Platform:', platform.platform())
```

```python
packages = ['numpy', 'matplotlib', 'ipykernel', 'jupyter']
for package in packages:
    try:
        print(f'{package}: {importlib.metadata.version(package)}')
    except importlib.metadata.PackageNotFoundError:
        print(f'{package}: NOT INSTALLED')
```

```python
# Optional: verify that NumPy and Matplotlib can be imported.
try:
    import numpy as np
    import matplotlib
    print('NumPy import successful; example array:', np.array([1, 2, 3]))
    print('Matplotlib import successful; version:', matplotlib.__version__)
except ImportError as error:
    print('Import check failed:', error)
    print('Install dependencies with: pip install -r requirements.txt')
```
